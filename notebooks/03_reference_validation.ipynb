# Reference Image Validation

## Objective

Visual/statistical spot-check of LDCT-IQAC reference images beyond the automated corruption/duplicate checks already run by `scripts/audit_dataset.py` and `scripts/validate_dataset.py` -- a human-in-the-loop pass, not a re-implementation of those checks.

## Intended analysis

Sample references across the `expert_score` range and across image statistics outliers; display them; note anything the automated audit would not catch (e.g. subjectively unusual anatomy or windowing).
